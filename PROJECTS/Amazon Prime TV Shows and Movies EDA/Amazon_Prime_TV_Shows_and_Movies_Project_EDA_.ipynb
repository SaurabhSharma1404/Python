## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

''' we will replace seasons nan or missing value with zero '''
titles['seasons'] = titles['seasons'].fillna(0)
````

````python
''' filled the missing value in description column with not available '''

titles['description'] = titles['description'].fillna('Not available')
````

````python
''' filled missing values in age_certification column with not rated '''

titles['age_certification'] = titles['age_certification'].fillna('Not rated')
````

````python
''' droping the duplicate values from both the dataset '''

titles.drop_duplicates(inplace=True)
credits.drop_duplicates(inplace=True)
````

````python
''' here we write a function which rate the movie or show based on imdb score
into 4 category missing will be not rated and rest according to imdb score '''

def rating(x):
  if pd.isna(x):
    return 'Not Rated'
  elif x<5:
    return 'Low Rated'
  elif x<7:
    return 'Medium Rated'
  else:
    return 'Highly Rated'

''' look here we created a new column Rating and apply the function rating '''

titles['Rating'] = titles['imdb_score'].apply(rating)
````

````python
''' here we have created new column runtime_category which categories
runtime with short, medium and long '''

def runtime(x):
  if pd.isna(x):
    return 'Not Available'
  elif x<60:
    return 'Short'
  elif x<=120:
    return 'Medium'
  else:
    return 'long'


# so here we applied function runtime and created new column

titles['Runtime_category'] = titles['runtime'].apply(runtime)
````

````python
''' here ast.literal function covert text string in column genres to
actual python list and the reason we coverted into python list it
helps us in searching, sorting and many others '''

import ast
titles['genres'] = titles['genres'].apply(
    lambda x: ast.literal_eval(x) if pd.notna(x) and x != '[]' else []
)
````

````python
''' we have done same here like genres column that we have coverted string to
actual python list which will we helpful to us '''

import ast

titles['production_countries'] = titles['production_countries'].apply(
    lambda x: ast.literal_eval(x) if pd.notna(x) and x.strip() != '[]' else []
)
````

````python
''' so here is the summary of missing or nan values in these 5 columns and we will
not going to replace them with median because pandas function skips nan values
and provide best results and the main reason we are not filling the nan values
is that they can create a artifical spike in charts and more  '''

missing_summary = pd.DataFrame({
    'Missing Count': titles[['imdb_id', 'imdb_score', 'imdb_votes', 'tmdb_popularity', 'tmdb_score']].isna().sum(),
    'Percentage (%)': (titles[['imdb_id', 'imdb_score', 'imdb_votes', 'tmdb_popularity', 'tmdb_score']].isna().mean() * 100).round(2)
})
print(missing_summary)
````

````python
''' now we will merge the two dataset on left join
we used left join because the main dataset is titles so we get all data from
titles and matching from credits '''


merged_datasets = pd.merge(titles,credits,how='left',on='id')



````

````python
''' some valuable insights out of merged_dataset'''

total_titles = merged_datasets.shape[0]

total_no_of_movies = merged_datasets[merged_datasets['type'] == 'MOVIE'].shape[0]

total_no_of_tvshows = merged_datasets[merged_datasets['type'] == 'SHOW'].shape[0]

top_1_production_country = (merged_datasets.explode('production_countries').groupby('production_countries').size().sort_values(ascending=False).head(1))

# here in above code we use explode to take out data stored in single cell to own seprate row
top_watch_genre = merged_datasets.explode("genres").groupby('genres').size().sort_values(ascending=False).head(1)

lowest_watched_genre = merged_datasets.explode('genres').groupby('genres').size().sort_values().head(1)

avg_runtime = merged_datasets['runtime'].mean().round(2)

avg_imdb_score = merged_datasets['imdb_score'].mean().round(2)

movie_count = ((merged_datasets['type'] == 'MOVIE') & merged_datasets['release_year'].between(2000,2010)).sum()
''' so in movie_count we get information of how many movies released between year 2000 to 2010'''

tv_shows_count = ((merged_datasets['type'] == 'SHOW')& merged_datasets['release_year'].between(2000,2010)).sum()
''' here we get tv_shows release from year 2000 to 2010 '''








print(f'Total number of titles : {total_titles}')
print(f'Total number of movies : {total_no_of_movies}')
print(f'Total number of TV-Shows : {total_no_of_tvshows}')
print()
print(f'Highest production country : {top_1_production_country}')
print()
print(f'Top watched genre is : {top_watch_genre} ')
print() # for creating space in between
print(f'Lowest watched genre is : {lowest_watched_genre}')
print()
print(f'Average runtime is : {avg_runtime}')
print(f'Average IMDB score is : {avg_imdb_score}')
print(f'Total number of Movies released between 2000 to 2010 is : {movie_count}')
print(f'Total number of TV-Shows released between 2000 to 2010 is : {tv_shows_count}')


````
