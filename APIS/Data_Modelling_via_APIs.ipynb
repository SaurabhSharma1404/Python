# API(Application Programming Interface)
An API (Application Programming Interface) is a toolbox 🛠️ that allows different software applications to communicate and share data with each other. It’s like a bridge 🌉 that connects two systems, enabling them to interact without needing to know how the other works internally.



**What is an API?**

- API as a waiter in the Restaurant

- Customer(The user) = orders the food
- The Chef(The Database/ Server) = Prepare the food
- The Waiter (API)

## GOAL: The user wants to check the weather in their city.

1. Frontend (WebPage/ Android Application)
- The User enters the city name and Clicks on Get weather Button.

2. API Request to weather server/ database.

3. This API Request will be processed and a response will be generated.

4. The Weather server sends the response to the API

5. The API delivers the response to the user(BY Udating in the Application)

### **HTTP Methods and CRUD Operations**
**HTTP Methods:**

1. **GET:** Retrieves data from the server.
2. **POST:** Sends data to the server to create a new resource.
3. **PUT:** Updates or replaces an entire resource.
4. **PATCH:** Updates specific fields of a resource.
5. **DELETE:** Deletes a resource from the server.

```python
# HTTP = Hypertext Transfer Protocol
# HTTPs= Secured Hypertext Transfer Protocol (Data is Encrypted 128K)
```

# **POST📤**
Purpose: Send data to the server to create a new resource.

Use Case: When you need to add new information, such as creating a new user, uploading a photo, or submitting a form.

**Example:**

Request: POST /users

Payload: {"name": "Alice", "age": 25}

Response: Confirmation of the new user creation ✅.

Analogy: Like mailing a package 📦—you’re sending something to be added.

```python
new_employee = {
        "id": 10,
        "name": "Nasim",
        "age": 30,
        "gender": "female",
        "department": "sales",
        "salary": 900000
    }
```

```python
updated_data= requests.post("http://4.186.26.27:8787/employees", json= new_employee)
updated_data.json()
```

```python
response = requests.get("http://4.186.26.27:8787/employees")
pd.json_normalize(response.json())
```

# **PUT 📝**
Purpose: Update an existing resource completely.

Use Case: When you need to modify an entire resource, such as updating a profile or replacing data.

```python
nasim_updated = {
        "id": 10,
        "name": "Nasim",
        "age": 30,
        "gender": "Male",
        "department": "sales",
        "salary": 1100000
    }
```

```python
updated_data = requests.put("http://4.186.26.27:8787/employees/10", json=nasim_updated )
```

```python
response = requests.get("http://4.186.26.27:8787/employees")
pd.json_normalize(response.json())
```

# **PATCH** 🩹
Purpose: Partially update an existing resource.

Use Case: When you need to modify only specific fields, such as changing just the username without altering other details.

Example:

Request: PATCH /users/123

Payload: {"name": "Alice B."}

Response: Updated user data 🎯.

Analogy: Like fixing a typo ✍️ in a document.

```python
# Update the age of nasim to 26Y
new_data = {
    "age":26
}
```

```python
requests.patch("http://4.186.26.27:8787/employees/10", json=new_data )
```

```python
response = requests.get("http://4.186.26.27:8787/employees")
pd.json_normalize(response.json())
```

```python
'''
.get , .post, .put, .patch

put = Full replacement
patch = Update a particular feature

eg:PUT = Replace he authecator file, Update the emplyee record completely

The Prob with patch:


Patch can be dangerous when

eg: Patch/employee/10
{
  'Salary':60000
}
'''
```

# **DELETE🗑️**
Purpose: Remove a resource from the server.

Use Case: When you need to delete information, such as removing a user or deleting a file.

Example:

Request: DELETE /users/123

Response: Confirmation of deletion 🛑.

Analogy: Like throwing something away 🗑️—it’s gone.

```python
# Delete the emplyee Nasim

requests.delete("http://4.186.26.27:8787/employees/10")
```
